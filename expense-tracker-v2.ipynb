My biggest expense is travel money.

```python
import pandas as pd
csv_text = """date,item,amount
2026-11-04,Food,100
2026-11-04,Travel,120
2026-11-04,Food,50
2026-11-04,Recharge,299
2026-11-04,Travel, 50
2026-11-05,Fun, 200
2026-11-05,Travel, 130
2026-11-05,Food, 50
2026-11-06,Travel, 130
2026-11-06,Stationery, 70
"""
with open("expenses.csv", "w") as f:
    f.write(csv_text)

df = pd.read_csv("expenses.csv")
print(df)
print(df.groupby("item")["amount"].sum())
totals = df.groupby("item")["amount"].sum().sort_values(ascending=False)
print(totals)
totals.plot(kind="pie",autopct="%1.0f%%",title="my daily expenses",ylabel=" ")
```
